# Benefits and Cost Sharing Data Processing
This notebook processes the `Benefits_Cost_Sharing_PUF.csv` dataset. It loads the dataset and renames the columns to be much more descriptive and user-friendly.

```python
import pandas as pd

# File path to the dataset
file_path = '../data_source/Benefits_Cost_Sharing_PUF.csv'

# Load the dataset (using low_memory=False because the dataset is quite large)
df = pd.read_csv(file_path, low_memory=False)

# Display the initial columns
print("Original Columns:")
print(df.columns.tolist())
```

```python
# Dictionary mapping old column names to new, descriptive column names
column_mapping = {
    'BusinessYear': 'Business Year',
    'StateCode': 'State Code',
    'IssuerId': 'Issuer ID',
    'SourceName': 'Source Name',
    'ImportDate': 'Import Date',
    'StandardComponentId': 'Standard Component ID',
    'PlanId': 'Plan ID',
    'BenefitName': 'Benefit Name',
    'CopayInnTier1': 'Copay In-Network Tier 1',
    'CopayInnTier2': 'Copay In-Network Tier 2',
    'CopayOutofNet': 'Copay Out-of-Network',
    'CoinsInnTier1': 'Coinsurance In-Network Tier 1',
    'CoinsInnTier2': 'Coinsurance In-Network Tier 2',
    'CoinsOutofNet': 'Coinsurance Out-of-Network',
    'IsEHB': 'Is Essential Health Benefit (EHB)',
    'IsCovered': 'Is Covered',
    'QuantLimitOnSvc': 'Quantitative Limit on Service',
    'LimitQty': 'Limit Quantity',
    'LimitUnit': 'Limit Unit',
    'Exclusions': 'Exclusions',
    'Explanation': 'Explanation',
    'EHBVarReason': 'EHB Variance Reason',
    'IsExclFromInnMOOP': 'Is Excluded from In-Network Max-Out-of-Pocket',
    'IsExclFromOonMOOP': 'Is Excluded from Out-of-Network Max-Out-of-Pocket'
}

# Rename the columns
df_renamed = df.rename(columns=column_mapping)

# Display the new columns
print("\nNew Columns:")
print(df_renamed.columns.tolist())
```

```python
# Preview the first few rows of the transformed dataset
df_renamed.head()
```

## Data Cleaning
Next, we perform standard data science cleaning steps: handling missing values, formatting data types, and standardizing strings.

```python
# 1. Drop duplicates
initial_shape = df_renamed.shape
df_cleaned = df_renamed.drop_duplicates()
print(f"Dropped {initial_shape[0] - df_cleaned.shape[0]} duplicate rows.")
```

```python
# 2. Convert Date columns to Datetime
if 'Import Date' in df_cleaned.columns:
    # The warnings about SettingWithCopyWarning can be avoided by ensuring we have a clean copy
    df_cleaned = df_cleaned.copy()
    df_cleaned['Import Date'] = pd.to_datetime(df_cleaned['Import Date'], errors='coerce')
    print("Converted 'Import Date' to datetime.")
```

```python
# 3. Clean numeric fields (like percentages: '35.00%' -> 0.35)
def clean_percentage(val):
    if isinstance(val, str) and '%' in val:
        try:
            return float(val.replace('%', '').strip()) / 100.0
        except:
            return val
    return val

percentage_cols = ['Coinsurance In-Network Tier 1', 'Coinsurance In-Network Tier 2', 'Coinsurance Out-of-Network']
for col in percentage_cols:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].apply(clean_percentage)
        # Also replace 'Not Applicable' with NaN for numeric representation
        df_cleaned[col] = df_cleaned[col].replace('Not Applicable', pd.NA)

print("Cleaned percentage columns.")
```

```python
# 4. Standardize text data (Categorical columns)
categorical_cols = ['Is Covered', 'Is Essential Health Benefit (EHB)', 'Quantitative Limit on Service']
for col in categorical_cols:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].astype(str).str.strip().str.title()
        df_cleaned[col] = df_cleaned[col].replace('Nan', pd.NA)

print("Standardized categorical columns.")
```

```python
# 5. Handling Missing Values
# Calculate missing percentage per column
missing_pct = df_cleaned.isnull().mean() * 100
print("Missing value percentages (> 0%):")
print(missing_pct[missing_pct > 0].sort_values(ascending=False))

# Optional: drop columns with more than 90% missing values
threshold = 90.0
cols_to_drop = missing_pct[missing_pct > threshold].index
df_cleaned = df_cleaned.drop(columns=cols_to_drop)
print(f"\nDropped columns with > {threshold}% missing data: {list(cols_to_drop)}")
```

## Exploratory Data Analysis (EDA) & Visualization
Now that the data is cleaned, the next step is Exploratory Data Analysis (EDA). Here, we will visualize the data to uncover underlying distributions, patterns, and relationships.

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Plotly provides interactive charts automatically!
```

```python
# 1. Distribution of Plans by State
if 'State Code' in df_cleaned.columns:
    state_counts = df_cleaned['State Code'].value_counts().head(20).reset_index()
    state_counts.columns = ['State Code', 'Count']
    fig = px.bar(state_counts, x='State Code', y='Count', 
                 title='Top 20 States by Number of Plan Benefits',
                 color='Count', color_continuous_scale='Viridis')
    fig.show()
```

```python
# 2. Coverage Status Distribution
if 'Is Covered' in df_cleaned.columns:
    cov_counts = df_cleaned['Is Covered'].value_counts().reset_index()
    cov_counts.columns = ['Is Covered Status', 'Frequency']
    fig = px.bar(cov_counts, x='Is Covered Status', y='Frequency', 
                 title='Distribution of Coverage Status', color='Is Covered Status',
                 color_discrete_sequence=px.colors.qualitative.Pastel)
    fig.show()
```

```python
# 3. Coinsurance In-Network Tier 1 Distribution
if 'Coinsurance In-Network Tier 1' in df_cleaned.columns:
    plot_data = pd.to_numeric(df_cleaned['Coinsurance In-Network Tier 1'], errors='coerce').dropna()
    fig = px.histogram(plot_data, nbins=20, 
                       title='Distribution of Coinsurance (In-Network Tier 1)',
                       labels={'value': 'Coinsurance Rate (e.g., 0.20 = 20%)'},
                       color_discrete_sequence=['teal'])
    fig.update_layout(showlegend=False, yaxis_title='Frequency')
    fig.show()
```

```python
# 4. Essential Health Benefit (EHB) vs. Coverage
if 'Is Essential Health Benefit (EHB)' in df_cleaned.columns and 'Is Covered' in df_cleaned.columns:
    ehb_cov = df_cleaned.groupby(['Is Essential Health Benefit (EHB)', 'Is Covered']).size().reset_index(name='Count')
    fig = px.bar(ehb_cov, x='Is Essential Health Benefit (EHB)', y='Count', color='Is Covered',
                 barmode='group', title='Coverage Status by Essential Health Benefit (EHB) Classification',
                 color_discrete_sequence=px.colors.qualitative.Set2)
    fig.show()
```

## Advanced EDA: Insurance Coverage Analysis
Can we know what insurance covers which healthcare services? Yes! By analyzing \Issuer ID\, \Plan ID\, \Benefit Name\, and \Is Covered\, we can map out exactly how insurance companies cover different healthcare needs.

```python
# 5. Most Covered vs. Least Covered Healthcare Services (Benefits)
if 'Benefit Name' in df_cleaned.columns and 'Is Covered' in df_cleaned.columns:
    df_cleaned['Is Covered Binary'] = (df_cleaned['Is Covered'] == 'Covered').astype(int)
    benefit_coverage = df_cleaned.groupby('Benefit Name')['Is Covered Binary'].agg(['mean', 'count']).reset_index()
    benefit_coverage = benefit_coverage[benefit_coverage['count'] > 100]
    
    top_10 = benefit_coverage.sort_values('mean', ascending=False).head(10)
    top_10['mean'] = top_10['mean'] * 100
    
    bottom_10 = benefit_coverage.sort_values('mean', ascending=True).head(10)
    bottom_10['mean'] = bottom_10['mean'] * 100
    
    fig = make_subplots(rows=1, cols=2, subplot_titles=('Top 10 Most Consistently Covered', 'Top 10 Least Covered'))
    
    fig.add_trace(go.Bar(y=top_10['Benefit Name'], x=top_10['mean'], orientation='h', marker_color='#2ca02c', name='Most Covered'), row=1, col=1)
    fig.add_trace(go.Bar(y=bottom_10['Benefit Name'], x=bottom_10['mean'], orientation='h', marker_color='#d62728', name='Least Covered'), row=1, col=2)
    
    fig.update_layout(title_text='Most vs Least Covered Healthcare Services', showlegend=False, height=500)
    fig.update_yaxes(autorange="reversed") # to show the highest on top
    fig.show()
```

```python
# 6. Insurance Companies (Issuers) with Highest Coverage Rates
if 'Issuer ID' in df_cleaned.columns and 'Is Covered Binary' in df_cleaned.columns:
    issuer_coverage = df_cleaned.groupby('Issuer ID')['Is Covered Binary'].agg(['mean', 'count']).reset_index()
    issuer_coverage = issuer_coverage[issuer_coverage['count'] > 1000]
    top_issuers = issuer_coverage.sort_values('mean', ascending=False).head(15)
    top_issuers['Issuer ID'] = top_issuers['Issuer ID'].astype(str)
    top_issuers['Coverage Rate (%)'] = top_issuers['mean'] * 100
    
    fig = px.bar(top_issuers, x='Issuer ID', y='Coverage Rate (%)', 
                 title='Top 15 Insurance Issuers by Overall Coverage Rate',
                 color='Coverage Rate (%)', color_continuous_scale='Blues')
    fig.update_layout(yaxis_range=[0, 105])
    fig.show()
```

```python
# 7. Average Coinsurance (Patient Cost Burden) by Top 15 Benefits
if 'Coinsurance In-Network Tier 1' in df_cleaned.columns and 'Benefit Name' in df_cleaned.columns:
    top_benefits = df_cleaned['Benefit Name'].value_counts().head(15).index
    plot_data = df_cleaned[df_cleaned['Benefit Name'].isin(top_benefits)].copy()
    plot_data['Coinsurance Numeric'] = pd.to_numeric(plot_data['Coinsurance In-Network Tier 1'], errors='coerce')
    
    fig = px.box(plot_data, y='Benefit Name', x='Coinsurance Numeric', color='Benefit Name',
                 title='Patient Coinsurance Burden by Top 15 Healthcare Services',
                 labels={'Coinsurance Numeric': 'Coinsurance Rate (e.g. 0.20 = 20% patient pays)'})
    fig.update_layout(showlegend=False, height=600)
    fig.show()
```

```python
# 8. Heatmap: State vs Top Benefits Coverage Rate
if 'State Code' in df_cleaned.columns and 'Benefit Name' in df_cleaned.columns:
    top_states = df_cleaned['State Code'].value_counts().head(15).index
    top_benefits = df_cleaned['Benefit Name'].value_counts().head(15).index
    heat_df = df_cleaned[(df_cleaned['State Code'].isin(top_states)) & (df_cleaned['Benefit Name'].isin(top_benefits))]
    
    pivot_table = heat_df.pivot_table(index='Benefit Name', columns='State Code', values='Is Covered Binary', aggfunc='mean') * 100
    
    fig = px.imshow(pivot_table, text_auto=".0f", aspect="auto",
                    color_continuous_scale='YlGnBu',
                    title='Coverage Rate (%) of Top Healthcare Services by Top States',
                    labels=dict(color="Coverage (%)", x="State Code", y="Healthcare Service"))
    fig.update_layout(height=600)
    fig.show()
```
